8. Create a function (using Biopython or without it) that computes the reverse complement of a given DNA sequence. Ensure that the input sequence is validated to contain only valid nucleotide characters (A, T, C, G).
Provide the output result fot the first 15 characters of the sequence.

(3 Points)

```python
def reverse_complement(dna):
    dna = str(dna).upper()

    # Checking if the DNA contains only A, T, C, G
    for base in dna:
        if base not in "ATCG":
            raise ValueError("Invalid DNA sequence!")

    # Creating complementary sequence
    complement = {
        "A": "T",
        "T": "A",
        "C": "G",
        "G": "C"
    }

    complementary_dna = ""

    for base in dna:
        complementary_dna += complement[base]

    # Reversing complementary sequence
    return complementary_dna[::-1]


# Applying the function to the first 15 nucleotides
dna_15 = record.seq[:15]

print("Original DNA:", dna_15)
print("Reverse complement:", reverse_complement(dna_15))
```

9. Write a function that takes your DNA sequence (first 21 characters) as input and performs both transcription and translation on a given DNA sequence. The function should:
- Transcribe the DNA sequence into mRNA. In mRNA, thymine (T) is replaced by uracil (U).
- Translate the mRNA sequence into a protein sequence using the standard genetic code. Do not forget about stop codons (they stop the translation).

(4 Points)

```python
from Bio.Seq import Seq

def transcribe_translate(dna):
    dna = str(dna).upper()

    # Checking invalid nucleotides
    for base in dna:
        if base not in "ATCG":
            raise ValueError("Invalid DNA sequence!")

    # Converting DNA into a Biopython Seq object
    dna_seq = Seq(dna)

    # Transcribing DNA into mRNA
    mrna = dna_seq.transcribe()

    # Translating mRNA into protein
    protein = mrna.translate(to_stop=True)

    return mrna, protein


# Use the first 21 nucleotides
dna_21 = record.seq[:21]

mrna, protein = transcribe_translate(dna_21)

print("DNA sequence:", dna_21)
print("mRNA sequence:", mrna)
print("Protein sequence:", protein)
```

10. Write a function that calculates the total number of single-nucleotide substitutions (SNPs) in all possible codons (except STOP codons) that can result in a stop codon (TAA, TAG, or TGA).  Use Standard RNA codon table from [wikipedia](https://en.wikipedia.org/wiki/DNA_and_RNA_codon_tables).  
Hints:  The simplest way to solve this task is to iterate over all possible codons and check how many single nucleotide changes (mutations) transform a sense codon into a stop codon.

(4 Points)

```python
from itertools import product

def count_stop_mutations():

    nucleotides = "AUCG"
    stop_codons = {"UAA", "UAG", "UGA"}

    # Generating all possible RNA codons
    codons = [
        "".join(codon)
        for codon in product(nucleotides, repeat=3)
    ]

    count = 0

    # Go through each codon
    for codon in codons:

        # Exclude stop codons
        if codon in stop_codons:
            continue

        # Try changing each of the 3 positions
        for i in range(3):

            for nucleotide in nucleotides:

                # Only count actual substitutions
                if nucleotide != codon[i]:

                    mutated_codon = (
                        codon[:i]
                        + nucleotide
                        + codon[i+1:]
                    )

                    # Check whether mutation creates STOP
                    if mutated_codon in stop_codons:
                        count += 1

    return count


result = count_stop_mutations()

print("Total SNPs resulting in STOP codons:", result)
```
